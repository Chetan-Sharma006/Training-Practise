```python
from sklearn.preprocessing import StandardScaler
cols=['age','bmi','children']
scaler = StandardScaler()
df_cleaned[cols]= scaler.fit_transform(df_cleaned[cols])
df_cleaned.head()
```

```python
from scipy.stats import pearsonr
selected_features=['age','sex','bmi','children','smoker','charges',	'region_northwest','region_southeast','region_southwest','bmi_categories_Normal','bmi_categories_Overweight','bmi_categories_Obese']
```

```python
correlations={
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
```

```python
correlation_df=pd.DataFrame(list(correlations.items()),columns=['feature','Pearsonr correlation'])
correlation_df.sort_values(by='Pearsonr correlation',ascending=False)
```
